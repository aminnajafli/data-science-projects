```python
import pandas as pd 
cars=pd.read_excel(r"C:\Users\Amin\Downloads\cars.xlsx")
cars
```

```python
#relationship between highway and city miles per gallon 
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(cars['cty'], cars['hwy'], color='red', alpha=0.7)
plt.title('relationship between city and highway miles per gallon')
plt.xlabel('city MPG (cty)')
plt.ylabel('highway MPG (hwy)')
plt.grid(True)
plt.show()
```

```python
"""there is a strong positive linear correlation between city (cty) and highway (hwy) miles per gallon. 
as a vehicle's city fuel efficiency increases, its highway fuel efficiency increases proportionally."""
```

```python
#pairwise scatter plots and distributions grouped by drive train
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(cars, hue='drv')
plt.show()
```

```python
"""highest correlation: cty vs hwy (city MPG and highway MPG). 
the scatter plot displays a tight, narrow upward diagonal pattern indicating a near-perfect positive linear relationship.
lowest correlation: year vs any numerical variable (e.g., year vs cty, year vs displ, or year vs cyl).
the data points form parallel vertical/horizontal lines (since years are discrete, mostly 1999 and 2008), 
showing virtually zero linear correlation."""
```

```python
#moat frequent car type
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.countplot(data=cars, x='class', order=cars['class'].value_counts().index)
plt.title('frequency of each car type (class)')
plt.xlabel('car type (class)')
plt.ylabel('count')
plt.xticks(rotation=45)
plt.show()

print(cars['class'].value_counts())
```

```python
#cylinders and drive train frequency 
import seaborn as sns
import matplotlib.pyplot as plt

# bar plot showing average/total cylinders for each drive train
plt.figure(figsize=(8, 5))
sns.barplot(data=cars, x='drv', y='cyl', ci=None)
plt.title('number of cylinders by drive train')
plt.xlabel('drive train (f = FWD, r = RWD, 4 = 4WD)')
plt.ylabel('cylinders')
plt.show()

# count plot to determine the most frequent drive train
plt.figure(figsize=(8, 5))
sns.countplot(data=cars, x='drv', order=cars['drv'].value_counts().index)
plt.title('frequency of each drive train')
plt.xlabel('drive train')
plt.ylabel('count')
plt.show()

print(cars['drv'].value_counts())
```

```python
#engine displacement by class 
import seaborn as sns
import matplotlib.pyplot as plt

#boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(data=cars, x='class', y='displ')
plt.title('engine displacement by car class (boxplot)')
plt.xlabel('car class')
plt.ylabel('engine displacement (L)')
plt.xticks(rotation=45)
plt.show()

#violinplot
plt.figure(figsize=(10, 6))
sns.violinplot(data=cars, x='class', y='displ')
plt.title('engine displacement by car class (violinplot)')
plt.xlabel('car class')
plt.ylabel('engine displacement (L)')
plt.xticks(rotation=45)
plt.show()
```
