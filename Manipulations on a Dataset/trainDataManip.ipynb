```python
#feature correlations(var_0 to var_9) 
features=[f'var_{i}' for i in range(10)] 

#find correlations in features 
corr_matrix=df[features].corr() 

#sort most correlated pairs (excluding self-correlations) 
corr_pairs=corr_matrix.unstack() 
sorted_corr= corr_pairs[corr_pairs.index.get_level_values(0) !=corr_pairs.index.get_level_values(1)] 
sorted_corr=sorted_corr.abs().sort_values(ascending=False) 

#show top 10 most correlated unique pairs 
top_10_corr=sorted_corr.iloc[::2].head(10) 
print(top_10_corr)
```

```python
#remove outliers using Interquartile Range (IQR) method
q1=df.quantile(0.25) 
q3=df.quantile(0.75) 
iqr=q3-q1 

df_no_outliers=df[~((df<(q1-1.5*iqr)) | (df> (q3+1.5*iqr))).any(axis=1)]
df_no_outliers
```
