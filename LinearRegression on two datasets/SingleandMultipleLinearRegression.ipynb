```python
def outlier_deterction(cols): 
    q1, q3=np.percentile(cols,[25,75]) 
    iqr=q3-q1 
    upper_bound=q3+(1.5*iqr) 
    lower_bound=q1-(1.5*iqr) 
    return upper_bound,lower_bound
```

```python
list1=['MinTemp', 'MaxTemp'] 
for cols in list1: 
    upper_bound, lower_bound=outlier_deterction(dataset[cols]) 
    dataset[cols]= np.clip(dataset[cols], a_min=lower_bound, a_max=upper_bound) 
    
```

```python
dataset.plot(x='MinTemp', y='MaxTemp', style='o') 
plt.title('MinTemp vs MaxTemp') 
plt.xlabel('MinTemp') 
plt.ylabel('MaxTemp') 
plt.show()
```
